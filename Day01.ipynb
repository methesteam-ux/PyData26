### 리스트와 반복구조 연습(p24)

```python
def getIndex(num_list:list, target):
    for i in range(0, len(num_list)-1):
        if num_list[i] == target:
            return i
    return -1

def getMax(num_list:list):
    maxnum = num_list[0]
    for element in num_list:
        if element > maxnum:
            maxnum = element
    return maxnum

def getMin(num_list:list):
    minnum = num_list[0]
    for element in num_list:
        if element < minnum:
            minnum = element
    return minnum

def countGT(num_list:list, target:int):
    sum = 0
    for element in num_list:
        if element > target:
            sum += 1
    return sum

def sumList(num_list:list):
    sum = 0
    for element in num_list:
        sum += element
    return sum

def swapList(num_list:list):
    temp_list = []
    for element in num_list:
        temp_list.append(element)

    i1 = 0
    for i2 in range(-1, -1-len(temp_list), -1):
        num_list[i1] = temp_list[i2]
        i1 += 1
```

```python
number_list = [23, 45, 27, 11, 25, 65, 78]

print(getIndex(number_list, 25))
print(getMax(number_list))
print(getMin(number_list))
print(countGT(number_list, 42))
print(sumList(number_list))
swapList(number_list)
print(number_list)
```

### [실습] 모스 부호 번역기(p51)

```python
morse_dict = {
    "a": ".-", 
    "b": "-...", 
    "c": "-.-.", 
    "d": "-..", 
    "e": ".", 
    "f": "..-.", 
    "g": "--.", 
    "h": "....", 
    "i": "..", 
    "j": ".---", 
    "k": "-.-", 
    "l": ".-..", 
    "m": "--", 
    "n": "-.", 
    "o": "---", 
    "p": ".--.", 
    "q": "--.-", 
    "r": ".-.", 
    "s": "...", 
    "t": "-", 
    "u": "..-", 
    "v": "...-", 
    "w": ".--", 
    "x": "-..-", 
    "y": "-.--", 
    "z": "--..",
}

def translate_morse(string:str):
    for i in range(len(string)):
        code = ord(string[i])

        if code < 65 or code > 122:
            return -1

    str_low = string.lower()

    morse_temp = []
    for i in range(len(str_low)):
        morse_temp.append(morse_dict[str_low[i]])

    string_morse = " ".join(morse_temp)
    return string_morse
```

```python
string = input("> 알파벳 문자열 입력 :")
print(">> 모스 부호 : " + translate_morse(string))
```
